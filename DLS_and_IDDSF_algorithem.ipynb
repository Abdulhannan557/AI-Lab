```python

import networkx as nx
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from collections import deque

# --------------------------------
# 1. SEARCH TREE
# --------------------------------

tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H', 'I'],
    'E': ['J'],
    'F': ['K'],
    'G': [],
    'H': [], 'I': [], 'J': [], 'K': []
}


# --------------------------------
# 2. DEPTH LIMITED SEARCH (DLS)
# --------------------------------

def dls(node, goal, limit, visited_order):

    visited_order.append(node)

    if node == goal:
        return 'found'

    if limit == 0:
        return 'cutoff'

    cutoff_occurred = False

    for child in tree[node]:

        res = dls(child, goal, limit - 1, visited_order)

        if res == 'found':
            return 'found'

        if res == 'cutoff':
            cutoff_occurred = True

    return 'cutoff' if cutoff_occurred else 'failure'


# --------------------------------
# 3. CREATE GRAPH AND TREE LAYOUT
# --------------------------------

G = nx.DiGraph([
    (parent, child)
    for parent, children in tree.items()
    for child in children
])

pos, next_x = {}, [0]


def layout(node, depth=0):

    children = tree[node]

    if not children:
        pos[node] = (next_x[0], -depth)
        next_x[0] += 1

    else:
        for child in children:
            layout(child, depth + 1)

        pos[node] = (
            sum(pos[child][0] for child in children) / len(children),
            -depth
        )


layout('A')


# --------------------------------
# 4. VISUALIZE EACH DEPTH ITERATION
# --------------------------------

def save_search_image(visited_order, depth):

    visited_nodes = set(visited_order)

    # Green = visited, Grey = unvisited
    colors = [
        'green' if node in visited_nodes else 'lightgrey'
        for node in G.nodes
    ]

    plt.figure(figsize=(8, 5))

    nx.draw(
        G,
        pos,
        with_labels=True,
        node_color=colors,
        node_size=900,
        edgecolors='black',
        font_weight='bold',
        font_color='black',
        arrows=False
    )

    plt.title(f"IDDFS Search - Depth Limit {depth}")

    filename = f"iddfs_depth_{depth}.png"

    plt.savefig(
        filename,
        dpi=130,
        bbox_inches='tight'
    )

    plt.close()

    print("Image saved:", filename)


# --------------------------------
# 5. ITERATIVE DEEPENING DFS (IDDFS)
# --------------------------------

def iddfs(start, goal, max_depth=10):

    total_visited = 0

    print("\n========== IDDFS SEARCH ==========")

    for depth in range(max_depth + 1):

        order = []

        res = dls(start, goal, depth, order)

        total_visited += len(order)

        print(
            f"Depth {depth}: {' '.join(order)}"
            f" -> {res} | Visited: {len(order)}"
        )

        # Save image for every depth limit
        save_search_image(order, depth)

        if res == 'found':

            print("\nGoal found at depth:", depth)
            print("Total IDDFS nodes visited:", total_visited)

            return depth, total_visited

        if res == 'failure':
            return None, total_visited

    return None, total_visited


# --------------------------------
# 6. BREADTH FIRST SEARCH (BFS)
# --------------------------------

def bfs(start, goal):

    queue = deque([start])

    visited = set()
    visited_order = []

    while queue:

        node = queue.popleft()

        if node in visited:
            continue

        visited.add(node)
        visited_order.append(node)

        if node == goal:
            break

        for child in tree[node]:

            if child not in visited:
                queue.append(child)

    print("\n========== BFS SEARCH ==========")

    print("Visited order:", ' '.join(visited_order))
    print("Total BFS nodes visited:", len(visited_order))

    return len(visited_order)


# --------------------------------
# 7. RUN AND COMPARE BOTH ALGORITHMS
# --------------------------------

goal = 'J'

depth, iddfs_count = iddfs('A', goal)

bfs_count = bfs('A', goal)

print("\n========== FINAL COMPARISON ==========")

print("Goal node:", goal)
print("IDDFS nodes visited:", iddfs_count)
print("BFS nodes visited:", bfs_count)

if iddfs_count > bfs_count:

    print("BFS visited fewer nodes.")

elif bfs_count > iddfs_count:

    print("IDDFS visited fewer nodes.")

else:

    print("Both visited the same number of nodes.")


# --------------------------------
# 8. FINAL TREE VISUALIZATION
# --------------------------------

NAVY = '#0E367B'

colors = [
    NAVY if node == 'J' else 'white'
    for node in G.nodes
]

plt.figure(figsize=(8, 5))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color=colors,
    node_size=900,
    edgecolors=NAVY,
    edge_color=NAVY,
    font_weight='bold',
    font_color='black',
    arrows=False
)

plt.title("Search Tree - Goal J Highlighted")

plt.savefig(
    'lab04_tree.png',
    dpi=130,
    bbox_inches='tight'
)

plt.close()

print("\nFinal tree saved: lab04_tree.png")
print("All tasks completed successfully!")
```

```python

# 8-Puzzle using Iterative Deepening DFS (IDDFS)

# Start and Goal states
START = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


# --------------------------------
# Generate possible moves
# --------------------------------

def actions(state):

    moves = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    # Up
    if row > 0:
        moves.append(("Up", blank - 3))

    # Down
    if row < 2:
        moves.append(("Down", blank + 3))

    # Left
    if col > 0:
        moves.append(("Left", blank - 1))

    # Right
    if col < 2:
        moves.append(("Right", blank + 1))

    return moves


# --------------------------------
# Generate new state
# --------------------------------

def result(state, new_position):

    blank = state.index(0)

    new_state = list(state)

    new_state[blank], new_state[new_position] = (
        new_state[new_position], new_state[blank]
    )

    return tuple(new_state)


# --------------------------------
# Depth Limited Search
# --------------------------------

def dls(state, limit, path, expanded):

    # Check goal
    if state == GOAL:
        return True

    # Stop at depth limit
    if limit == 0:
        return False

    # Count expanded node
    expanded[0] += 1

    # Explore possible moves
    for move, new_position in actions(state):

        new_state = result(state, new_position)

        # Avoid revisiting states in current path
        if new_state not in path:

            path.append(new_state)

            found = dls(
                new_state,
                limit - 1,
                path,
                expanded
            )

            if found:
                return True

            path.pop()

    return False


# --------------------------------
# Iterative Deepening DFS
# --------------------------------

def iddfs(start, max_depth=20):

    total_expanded = 0

    print("========== IDDFS SEARCH ==========")

    for depth in range(max_depth + 1):

        path = [start]

        expanded = [0]

        found = dls(
            start,
            depth,
            path,
            expanded
        )

        total_expanded += expanded[0]

        print(
            f"Depth {depth}: "
            f"Nodes expanded = {expanded[0]}"
        )

        if found:

            print("\nGoal Found!")

            print("Depth:", depth)

            print("Moves:")

            for state in path:
                print(state[0:3])
                print(state[3:6])
                print(state[6:9])
                print()

            print("Total nodes expanded:", total_expanded)

            return depth, total_expanded

    print("Goal not found.")

    return None, total_expanded


# --------------------------------
# Run IDDFS
# --------------------------------

depth, total = iddfs(START)

print("\n========== FINAL RESULT ==========")
print("Goal depth:", depth)
print("Total nodes expanded:", total)
```

```python

# Depth Limited Search (DLS) on Lab 3 Maze

# Maze
maze = [
    ['S', '.', '.', '.', '.', '#', '#'],
    ['#', '#', '.', '#', '#', '.', '.'],
    ['.', '.', '.', '#', '.', '.', '#'],
    ['.', '#', '.', '.', '.', '.', '#'],
    ['.', '.', '.', '.', '#', '.', 'G']
]

# Start and Goal
start = (0, 0)
goal = (4, 6)

# Movement order: Up, Right, Down, Left
directions = [
    (-1, 0),
    (0, 1),
    (1, 0),
    (0, -1)
]


# --------------------------------
# DEPTH LIMITED SEARCH
# --------------------------------

def dls(position, limit, path, visited):

    visited.append(position)

    # Check goal
    if position == goal:
        return 'found'

    # Check depth limit
    if limit == 0:
        return 'cutoff'

    cutoff_occurred = False

    row, col = position

    # Explore all possible directions
    for dr, dc in directions:

        new_row = row + dr
        new_col = col + dc

        # Check maze boundaries
        if (0 <= new_row < len(maze) and
            0 <= new_col < len(maze[0])):

            # Check wall
            if maze[new_row][new_col] != '#':

                new_position = (new_row, new_col)

                # Avoid revisiting current path
                if new_position not in path:

                    path.append(new_position)

                    result = dls(
                        new_position,
                        limit - 1,
                        path,
                        visited
                    )

                    if result == 'found':
                        return 'found'

                    if result == 'cutoff':
                        cutoff_occurred = True

                    path.pop()

    if cutoff_occurred:
        return 'cutoff'

    return 'failure'


# --------------------------------
# TEST DIFFERENT DEPTH LIMITS
# --------------------------------

limits = [5, 10, 15]

for limit in limits:

    path = [start]
    visited = []

    result = dls(
        start,
        limit,
        path,
        visited
    )

    print("\n--------------------------")
    print("Depth Limit:", limit)
    print("Result:", result)
    print("Nodes Visited:", len(visited))

    if result == 'found':

        print("Goal Found at:", goal)
        print("Solution Path:", path)

    else:
        print("Goal not reached within this limit.")
```
