**Task 2 — Variables and data types**

```python
student_name = "Your Name"
lab_number = 1
marks = 8.5
is_learning_ai = True

print("Student:", student_name)
print("Lab number:", lab_number)
print("Marks:", marks)
print("Learning AI:", is_learning_ai)

print("\nData types:")
print(type(student_name))
print(type(lab_number))
print(type(marks))
print(type(is_learning_ai))
```

Explanation: Variables store values. Python determines a value's type, such as str (text), int (integer), float (decimal), or bool (True/False).

**Task 3 — Arithmetic operators**

```python
a = 12
b = 5

print('Addition:', a + b)
print('Subtraction:', a - b)
print('Multiplication:', a * b)
print('Division:', a / b)
print('Floor division:', a // b)
print('Remainder:', a % b)
```

Explanation: These operators perform common calculations. The remainder operator % is often useful for checking whether a number is even or odd.

**Task 4 — Simple decision using if and else**

```python
number = 7

if number % 2 == 0:
    print(number, 'is even')
else:
    print(number, 'is odd')
```

Explanation: Python checks the condition after if. If it is true, the first block runs; otherwise, the else block runs.
